```python
import os
import glob
import pandas as pd

base_path = "/content/data/CSV"
output_path = "/content/full_CICIoT2023.csv"

files = glob.glob(base_path + "/**/*.csv", recursive=True)

if os.path.exists(output_path):
    os.remove(output_path)

total_rows = 0

for i, file in enumerate(files, start=1):

    attack_type = os.path.basename(os.path.dirname(file))

    for chunk in pd.read_csv(file, chunksize=50000):

        chunk["Attack_Type"] = attack_type

        chunk.to_csv(
            output_path,
            mode="a",
            header=(total_rows == 0),
            index=False
        )

        total_rows += len(chunk)

    print(f"Completed {i}/{len(files)} files | Rows: {total_rows:,}")

print("\n========== DATASET LOADING COMPLETE ==========")
print("Files processed:", len(files))
print("Total rows:", f"{total_rows:,}")
print("Total columns:", 40)
print("Saved at:", output_path)
```

```python
import pandas as pd
import numpy as np

file_path = "/content/full_CICIoT2023.csv"
output_path = "/content/CICIoT2023_sample.csv"

sample_size = 1_000_000
chunk_size = 500_000

# Calculate proportional sample targets
class_counts = {
    "DDoS-ICMP_Flood": 7200501,
    "DDoS-UDP_Flood": 5412231,
    "DDoS-TCP_Flood": 4497649,
    "DDoS-PSHACK_FLOOD": 4094772,
    "DDoS-SYN_Flood": 4059179,
    "DDoS-RSTFINFLOOD": 4045279,
    "DDoS-SynonymousIP_Flood": 3598133,
    "DoS-UDP_Flood": 3072993,
    "DoS-TCP_Flood": 2671430,
    "DoS-SYN_Flood": 2028836,
    "Benign_Final": 1098191,
    "Mirai-greeth_flood": 991834,
    "Mirai-udpplain": 890574,
    "Mirai-greip_flood": 751646,
    "DDoS-ICMP_Fragmentation": 452490,
    "VulnerabilityScan": 373351,
    "MITM-ArpSpoofing": 307560,
    "DDoS-UDP_Fragmentation": 286925,
    "DDoS-ACK_Fragmentation": 285075,
    "DNS_Spoofing": 178898,
    "Recon-HostDiscovery": 134378,
    "Recon-OSScan": 98259,
    "Recon-PortScan": 82284,
    "DoS-HTTP_Flood": 71861,
    "DDoS-HTTP_Flood": 28790,
    "DDoS-SlowLoris": 23426,
    "DictionaryBruteForce": 13064,
    "BrowserHijacking": 5859,
    "CommandInjection": 5409,
    "SqlInjection": 5245,
    "XSS": 3846,
    "Backdoor_Malware": 3218,
    "Recon-PingSweep": 2262,
    "Uploading_Attack": 1252
}

total_rows = sum(class_counts.values())

targets = {
    label: max(1, round(count / total_rows * sample_size))
    for label, count in class_counts.items()
}

# Adjust rounding so target is exactly 1 million
difference = sample_size - sum(targets.values())
largest_class = max(targets, key=targets.get)
targets[largest_class] += difference

# Reservoir sampling within each class
reservoirs = {label: [] for label in targets}
seen = {label: 0 for label in targets}

rng = np.random.default_rng(42)

for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    for label, group in chunk.groupby("Attack_Type"):
        records = group.to_dict("records")
        k = targets[label]

        for record in records:
            seen[label] += 1

            if len(reservoirs[label]) < k:
                reservoirs[label].append(record)
            else:
                j = rng.integers(0, seen[label])
                if j < k:
                    reservoirs[label][j] = record

    print("Processed rows:", sum(seen.values()))

# Combine sampled records
sample_parts = [
    pd.DataFrame(records)
    for records in reservoirs.values()
]

df_sample = pd.concat(sample_parts, ignore_index=True)
df_sample = df_sample.sample(frac=1, random_state=42).reset_index(drop=True)

df_sample.to_csv(output_path, index=False)

print("\nSample created successfully!")
print("Sample shape:", df_sample.shape)
print("Saved at:", output_path)
print("\nRecords per class:")
print(df_sample["Attack_Type"].value_counts())
```

```python
# Create binary target column
df_sample["Binary_Label"] = df_sample["Attack_Type"].apply(
    lambda x: "Benign" if x == "Benign_Final" else "Attack"
)

# Check class distribution
print("Binary class distribution:")
print(df_sample["Binary_Label"].value_counts())

print("\nDataset shape:", df_sample.shape)

# Save updated dataset
df_sample.to_csv("/content/CICIoT2023_sample.csv", index=False)

print("\nBinary labels created and dataset saved!")
```

```python
import numpy as np
import pandas as pd

# Separate features and labels
X = df_sample.drop(columns=["Attack_Type", "Binary_Label"])
y = df_sample[["Attack_Type", "Binary_Label"]]

# Replace infinite values with NaN
X = X.replace([np.inf, -np.inf], np.nan)

# Fill missing values with column medians
X = X.fillna(X.median())

# Reattach labels
df_clean = pd.concat([X, y], axis=1)

# Remove exact duplicate rows, including labels
df_clean = df_clean.drop_duplicates()

# Reset index
df_clean = df_clean.reset_index(drop=True)

print("Original dataset shape:", df_sample.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values:", df_clean.isnull().sum().sum())
print("Infinite values:", np.isinf(
    df_clean.select_dtypes(include=np.number)
).sum().sum())

print("\nBinary class distribution:")
print(df_clean["Binary_Label"].value_counts())
```

```python
# Select only the 39 original features
feature_cols = [
    col for col in df_clean.columns
    if col not in ["Attack_Type", "Binary_Label"]
]

# Create a hash for each feature record
df_clean["Feature_Hash"] = pd.util.hash_pandas_object(
    df_clean[feature_cols],
    index=False
)

# Check whether identical features have different attack labels
label_counts = df_clean.groupby("Feature_Hash")["Attack_Type"].nunique()

conflicts = (label_counts > 1).sum()

print("Feature records with conflicting labels:", conflicts)

# Remove temporary column
df_clean.drop(columns=["Feature_Hash"], inplace=True)
```

```python
feature_cols = [
    col for col in df_clean.columns
    if col not in ["Attack_Type", "Binary_Label"]
]

# Find feature combinations with multiple attack labels
conflicting_groups = (
    df_clean.groupby(feature_cols, dropna=False)["Attack_Type"]
    .nunique()
)

print("Total feature combinations:", len(conflicting_groups))
print("Conflicting combinations:", (conflicting_groups > 1).sum())

# Check whether these conflicts also affect binary labels
binary_conflicts = (
    df_clean.groupby(feature_cols, dropna=False)["Binary_Label"]
    .nunique()
)

print("Conflicts between Benign and Attack:", (binary_conflicts > 1).sum())
```
